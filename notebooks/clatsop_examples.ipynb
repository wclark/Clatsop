# Clatsop visual rule inspection

This notebook stays at the public game-model level: `Turn` objects represent board states, `Rule` objects represent primitive transformations, `Transition` objects represent completed turns, and a `Ruleset` finds and displays the legal choices. Internal bitmask storage is intentionally not part of this walkthrough.

```python
%matplotlib inline

from clatsop import Ruleset, Turn

ruleset = Ruleset.american()
initial_turn = Turn.initial()

print(f"Loaded {len(ruleset)} primitive American-checkers rules.")
```

## Starting state

A `Turn` displays its board and identifies the side to move.

```python
_ = initial_turn.display(size=4.2, title="Initial position: black to move")
```

## Legal rules for a state

Ask the ruleset which primitive rules are legal in the current turn, then inspect those rules visually. Each row shows the required board pattern, the resulting board pattern, and the move metadata.

```python
opening_rules = ruleset.legal_rules(initial_turn)
print(f"The initial position has {len(opening_rules)} legal rules.")
print(f"Capture forced: {ruleset.has_capture(initial_turn)}")
print(f"Promotion available: {ruleset.has_promotion(initial_turn)}")
_ = ruleset.display(
    opening_rules,
    size=3.2,
    title="Legal from the initial position",
)
```

## Apply a completed turn

`legal_transitions()` applies mandatory-capture policy and completes any capture chain before returning the next `Turn`.

```python
opening_transition = ruleset.legal_transitions(initial_turn)[0]
_ = opening_transition.turn.display(size=4.2, title="Resulting turn: white to move")
```

## Filtered and sorted rule views

Select rule objects for analysis or apply the same filters directly to the visual display. Here the promotion rules are grouped by side to move and capture status.

```python
promotion_rules = ruleset.rules_for(
    promotion=True,
    sort_by=("black_to_move", "capture"),
)
print(f"The ruleset has {len(promotion_rules)} promotion rules.")
_ = ruleset.display(
    promotion=True,
    sort_by=("black_to_move", "capture"),
    size=3.2,
    title="Promotion rules",
)
```

## Full visual rule catalog

The following output contains every primitive rule in the American ruleset. This is the exhaustive visual inspection surface: 366 separate condition/effect/summary rows, grouped by side and move metadata while retaining stable rule labels.

```python
_ = ruleset.display(
    sort_by=("black_to_move", "king", "capture", "promotion", "rule_index"),
    size=3.2,
)
```
