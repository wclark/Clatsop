# Clatsop full visual rule catalog

Every primitive American-checkers rule, displayed as a condition board, an effect board, and a concise game-level summary. The rows are grouped by side and move metadata; internal mask values are intentionally omitted.

```python
%matplotlib inline

from clatsop import Ruleset

ruleset = Ruleset.american()
print(f"Displaying all {len(ruleset)} primitive rules.")
```

```python
_ = ruleset.display(
    sort_by=("black_to_move", "king", "promotion", "capture", "rule_index"),
    size=3.2,
)
```
